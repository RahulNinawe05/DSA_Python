### Brut Force TC -O(N)

```python
n = 2
for i in range(0, n+1):
    if n % 2 == 0:
        print("PRIME")
        break
    else:
        print("NOT PRIME")
        break
```

### Optimal 

```python
def isprime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0:
        return False
    if n % 3 == 0:
        return False

    i = 5
    while i * i <= n:
        if n % i == 0 or  n % (i + 2) == 0:
            return False
        i += 6
    return True
```
