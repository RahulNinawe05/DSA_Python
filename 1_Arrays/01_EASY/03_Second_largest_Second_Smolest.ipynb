```python
num = [93,86,45,89,3,1,2,68,90,92,56,78,34,23,89]

largest = float("-inf")
Second_largest = float("-inf")

Smallest = float("inf")
Second_Smallest = float("inf")

n = len(num)

for i in range(0, n):
    if num[i] > largest:
        Second_largest = largest
        largest = num[i]
    elif num[i] > Second_largest and num[i] != largest:
        Second_largest = num[i]

for i in range(0,n):
    if num[i] < Smallest:
        Second_Smallest = Smallest
        Smallest = num[i]

    elif num[i] < Second_Smallest and num[i] !=  Smallest:
        Second_Smallest = num[i]

print(Second_largest,Second_Smallest)
```
