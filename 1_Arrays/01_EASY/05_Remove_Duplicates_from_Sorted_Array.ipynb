### Brute Solution

```python
"""
- which Data-Structure Unique and Maintain Order - Dict

- insight The Dict 
    - 1st input key to add in Dict
    - if the 1st and 2nd input key same, then 2nd input key to modify by '0'
    - if the 3rd is Unique Key , first check in Dict after add it ... so on
    
"""

# TC - O(2N) ~ O(N)
# SC - 

nums = [1,1,2,3,4,6,3,6,7,8,3,5,8] 

n = len(nums) 

x = {} # O(N)

for i in range(0,n): # O(N)
    x[nums[i]] = 0

j = 0
for k in x:          # O(N)
    nums[j] = k
    j += 1
print(j)
```
