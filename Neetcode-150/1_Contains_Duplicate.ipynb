```python
nums = [1,2,3,3,4,6,5,6]
```

```python
# Brutforce (Nested Loop)
def duplicate(n):
    for i in range(0,len(n)):
        for j in range(i + 1, len(n)):
            if n[i] == n[j]:
                return True
    return False
print(duplicate(nums))
```

```python
# Better  (Sorting)
nums.sort()
def doc(nums):
    for i in range(0,len(nums)-1):
        if nums[i] == nums[i + 1]:
            return True
    return False
doc(nums=nums)
```

```python
# Optimal   (Hashig Map)
def hashing(n):
    seen = set()
    for i in range(0, len(n)):
        if n[i] in seen:
            return True
        seen.add(n[i])
    return False
hashing(n = nums)
```
