```python
s = "racecar"
t = "carrace"
```

```python
# Brutforce (Sorting)
def brut_Anagram(s,t):
    if len(s) != len(t):
        return False
    a = sorted(s)
    b = sorted(t)
    if a == b:
        return True
    return False
brut_Anagram(s=s,t=t)
```

```python
# Better (Hashig Map)
def better_Anagram(s,t):
    if len(s) != len(t):
        return False
    hashing_map = {}
    for char in s:
        hashing_map[char]= hashing_map.get(char,0) + 1 
    
    for char in t:
        hashing_map[char] = hashing_map.get(char,0) - 1

    for count in hashing_map.values():
        if count != 0:
            return False
    return True
        
better_Anagram(s=s,t=t)
```

```python
# Optimal  (Hashig Map)
def optimal_Anagram(s,t):
    map_s = {}
    map_t = {}

    if len(s) != len(t):
        return False
    for char in s:
        map_s[char] = map_s.get(char,0) + 1
    for  char in t:
        map_t[char] = map_t.get(char,0) + 1 # why '+1' -> for Comparison

    if map_s == map_t:
        return True
    return False

optimal_Anagram(s=s,t=t)
```
