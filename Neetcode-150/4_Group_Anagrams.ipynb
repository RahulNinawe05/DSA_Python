```python
words = ["act","pots","tops","cat","stop","hat"]

hash_map = {}
for word in words:
    sorted_str = "".join(sorted(word))
    if sorted_str not in hash_map:
        hash_map[sorted_str] = []

    hash_map[sorted_str].append(word)

print(hash_map)
```
