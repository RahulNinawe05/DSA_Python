```python
# BrutForce (tc-O(n2)), (sc-O(1)) , nested loop



def brute_twosum(nums,target):
    for i in range(0,len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i] + nums[j] == target:
                return i , j

nums = [5,5,3,4]
target = 10
brute_twosum(nums=nums,target=target)
```

```python
# Optimal

def optimal_twosum(nums,target):
    hash_map = {}

    for i in range(0,len(nums)):
        complement = target - nums[i]
        if complement in hash_map:
            return hash_map[complement],i
        hash_map[nums[i]] = i

optimal_twosum(nums=nums,target=target)
```
