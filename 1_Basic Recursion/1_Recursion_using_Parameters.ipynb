```python
# print (x, n times)
def func(x,n):
    if n == 0:
        return 
    print(x)
    func(x,n-1)
```

```python
func(3,5)
```

```python
# print 1 to n Tail
def fun(n):
    if n == 0:
        return
    fun(n-1)
    print(n)
```

```python
fun(4)
```

```python
# print 1 to n haid
def fun(n):
    if n == 0:
        return
    print(n)
    fun(n-1)
```

```python
fun(3)
```

```python
def para(i,n):
    if i > n:
        return 
    print(i)
    para(i + 1, n)

para(1,4)
```

```python
def paa(i,n):
    if i > n:
        return
    paa(i+1,n)
    print(n - i)
paa(1,4)
```

```python
def fe(n):
    if n <= 0:
        return
    fe(n - 1)
    print(n)
fe(4)
```
