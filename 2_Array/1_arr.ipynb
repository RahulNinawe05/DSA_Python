```python
# Given an array arr[], check whether it is sorted in non-decreasing order. Return true if it is sorted otherwise false.

arr = [12,4445,64,75]

def sorted_value(n):
    for i in range(0,len(n) - 1):
        if n[i] > n[i + 1]:
            return False
    return True
print(sorted_value(arr))  
```
