```python
# find cout 
# Bubble sort karo aur total swaps count karo:
# pass(1)-
# [3,4,2,1]
# [3,2,4,1]
# [3,2,1,4]  # 4 is sort [4] 

# pass(2)-
# [3,2,1,4]
# [2,3,1,4]
# [2,1,3,4] # 3,4 is sort [3,4]

# pass(3)-
# [2,1,3,4]
# [1,2,3,4] # 2,3,4 is sort [2,3,4]

# pass(4)-
# [1,2,3,4] # 1 already sor [1,2,3,4]

nums3 = [1, 2, 3, 4, 5]     # [4, 3, 2, 1] 
n3= len(nums3)
count = 0
for i in range(n3):
    for j in range(0,n3-i-1):
        if nums3[j] > nums3[j+1]:
            nums3[j],nums3[j+1] = nums3[j+1],nums3[j]
            count+= 1
print(count)
```
