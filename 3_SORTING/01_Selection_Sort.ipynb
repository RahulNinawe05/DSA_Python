```python
def selection_sort(arr):
    n = len(arr)

    for i in range(n): # Index=4
        min_index = i # 0 = 5

        for j in range(i + 1,n):    # 0+1 = 1, Index=4
            if arr[j] < arr[min_index]: # 2 < 5 (yes)
                min_index = j   #  Index = 1

        arr[i], arr[min_index] = arr[min_index] , arr[i]    #( Index(0) = 5),(Index(1) = 2) = (Index(0) = 5),(Index(1) = 2) ==>  2 <=> 5 swip
    return arr


array = [5,9,3,10,45,2,0] 
print(selection_sort(array))
```

```python
# 6. Count Swaps Concept
# How many swaps are needed in Selection Sort for:

arr3 = [4, 3, 2, 1]

n3 = len(arr3)
count = 0

for i in range(n3):
    min_index3 = i
    for j in range(i+1,n3):
        if arr3[j] < arr3[min_index3]:
            min_index3 = j
            
    if min_index3 != i:
        arr3[i], arr3[min_index3] = arr3[min_index3] , arr3[i]
        count += 1

print(count)
```

```python
word = ["cat3 ant1 bat2"]
word = word[0].split()
lst_digit = [int(w[-1]) for w in word]

for w in range(len(lst_digit)):
    min_ind = w
    for j in range(w+1,len(lst_digit)):
        if lst_digit[j] < lst_digit[min_ind]:
            min_ind = j
    lst_digit[w], lst_digit[min_ind] = lst_digit[min_ind], lst_digit[w]
    word[w],word[min_ind]= word[min_ind],word[w]

word = list(map(lambda w: w[:-1], word))
print(word)
```
